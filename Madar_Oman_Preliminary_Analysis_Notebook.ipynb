```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Values transcribed from the project presentation.
# These are not independently recalculated from satellite imagery.
results = pd.DataFrame([
    {"site": "Al Khaburah", "event_area_km2": 0.88, "ordinary_day_area_km2": 0.41},
    {"site": "Saham",        "event_area_km2": 1.40, "ordinary_day_area_km2": 0.37},
    {"site": "Al Musannah",  "event_area_km2": 2.17, "ordinary_day_area_km2": 2.03},
])

results["difference_km2"] = results["event_area_km2"] - results["ordinary_day_area_km2"]
results["relative_difference_percent"] = (
    results["difference_km2"] / results["ordinary_day_area_km2"] * 100
)

results
```

## 2. Interpretation guide

The event-versus-ordinary-day difference is a comparison of the numbers reported in the presentation. A positive difference is not, by itself, proof that the entire area is floodwater. Radar responses can be affected by surface type, terrain, vegetation, viewing geometry, and acquisition timing.

```python
summary = results[["site", "event_area_km2", "ordinary_day_area_km2", "difference_km2"]].copy()
summary.columns = ["Site", "Event reported area (km²)", "Ordinary-day reported area (km²)", "Difference (km²)"]
summary
```

```python
ax = results.plot(
    x="site",
    y=["event_area_km2", "ordinary_day_area_km2"],
    kind="bar",
    figsize=(9, 5),
    rot=0
)
ax.set_title("Presentation-reported areas: event day vs ordinary day")
ax.set_xlabel("Study site")
ax.set_ylabel("Area (km²)")
ax.legend(["Event day", "Ordinary day"])
plt.tight_layout()
plt.show()
```

## 3. Preliminary findings

- **Al Khaburah:** the presentation reports 0.88 km² for the event comparison and 0.41 km² for an ordinary day, a difference of 0.47 km².
- **Saham:** the presentation reports 1.40 km² versus 0.37 km², a difference of 1.03 km².
- **Al Musannah:** the presentation reports 2.17 km² versus 2.03 km², a difference of 0.14 km². The presentation itself describes this case as not clearly distinguishable from the noise reference.

These are descriptive comparisons of reported values, not fresh measurements.
